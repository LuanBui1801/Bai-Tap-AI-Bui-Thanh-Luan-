```python
import numpy as np   #2.11
import skfuzzy as fuzz
from skfuzzy import control as ctrl
dist = ctrl.Antecedent(np.arange(0,50.1,0.1),'distance')
traffic = ctrl.Antecedent(np.arange(0,100.1,0.1),'traffic condition')
demand = ctrl.Antecedent(np.arange(0,100.1,0.1),'demand level')
weather = ctrl.Antecedent(np.arange(0,10.1,0.1),'weather condition')
rating = ctrl.Antecedent(np.arange(1,5.1,0.1),'customer rating')
time = ctrl.Antecedent(np.arange(0,100.1,0.1),'ride punctuality')
price = ctrl.Consequent(np.arange(0,100.1,0.1),'price')
bonus = ctrl.Consequent(np.arange(0,100.1,0.1),'Điểm thưởng')
dist['short'] = fuzz.trimf(dist.universe,[0,0,3])
dist['medium'] = fuzz.trimf(dist.universe,[2,5,8])
dist['long'] = fuzz.trimf(dist.universe,[6,13,20])
dist['very long'] = fuzz.trimf(dist.universe,[15,50,50])
traffic['low'] = fuzz.trimf(traffic.universe,[0,0,40])
traffic['medium'] = fuzz.trimf(traffic.universe,[20,50,80])
traffic['high'] = fuzz.trimf(traffic.universe,[60,100,100])
demand['low'] = fuzz.trimf(demand.universe,[0,0,30])
demand['medium'] = fuzz.trimf(demand.universe,[20,50,70])
demand['high'] = fuzz.trimf(demand.universe,[60,100,100])
weather['bad'] = fuzz.trimf(weather.universe,[0,0,4])
weather['moderate'] = fuzz.trimf(weather.universe,[3,5,7])
weather['good'] = fuzz.trimf(weather.universe,[6,10,10])
rating['poor'] = fuzz.trimf(rating.universe,[1,1,2.5])
rating['average'] = fuzz.trimf(rating.universe,[2,3,4])
rating['good'] = fuzz.trimf(rating.universe,[3.5,5,5])
time['late'] = fuzz.trimf(time.universe,[0,0,50])
time['on time'] = fuzz.trimf(time.universe,[40,60,80])
time['early'] = fuzz.trimf(time.universe,[70,100,100])
price['low'] = fuzz.trimf(price.universe,[0,0,40])
price['medium'] = fuzz.trimf(price.universe,[30,50,70])
price['high'] = fuzz.trimf(price.universe,[60,80,100])
price['very high'] = fuzz.trimf(price.universe,[85,100,100])
bonus['none'] = fuzz.trimf(bonus.universe,[0,0,20])
bonus['few'] = fuzz.trimf(bonus.universe,[10,30,50])
bonus['moderate'] = fuzz.trimf(bonus.universe,[40,60,80])
bonus['high'] = fuzz.trimf(bonus.universe,[70,100,100])
rules = [
    ctrl.Rule(dist['short'] & traffic['low'] & demand['low'], (price['low'])),
    ctrl.Rule(dist['short'] & traffic['medium'] & demand['high'], (price['medium'])),
    ctrl.Rule(dist['medium'] & traffic['high'] & demand['high'], price['high']),
    ctrl.Rule(dist['long'] & traffic['medium'] & weather['good'], (price['medium'])),
    ctrl.Rule(dist['long'] & traffic['high'] & weather['bad'], (price['very high'])),
    ctrl.Rule(dist['very long'] & traffic['high'] & demand['high'], (price['very high'])),
    ctrl.Rule(dist['medium'] & traffic['low'] & demand['low'], (price['medium'])),
    ctrl.Rule(dist['short'] & traffic['high'] & weather['bad'], (price['high'])),
    ctrl.Rule(dist['very long'] & weather['bad'], (price['very high'])),
    ctrl.Rule(dist['medium'] & traffic['medium'] & weather['moderate'], (price['medium'])),
    ctrl.Rule(rating['good'] & time['early'], (bonus['high'])),
    ctrl.Rule(rating['average'] & time['on time'], (bonus['moderate'])),
    ctrl.Rule(rating['poor'] & time['late'], (bonus['none'])),
    ctrl.Rule(dist['long'] & traffic['high'] & time['on time'], (bonus['high'])),
    ctrl.Rule(dist['medium'] & traffic['medium'] & rating['good'], (bonus['moderate'])),
    ctrl.Rule(rating['poor'] & time['late'], (bonus['none'])),
    ctrl.Rule(dist['very long'] & weather['bad'] & rating['good'], (bonus['high'])),
    ctrl.Rule(dist['short'] & rating['average'] & time['on time'], (bonus['few'])),
    ctrl.Rule(dist['long'] & traffic['high'] & time['late'], (bonus['few'])),
    ctrl.Rule(dist['medium'] & weather['moderate'] & rating['good'], (bonus['moderate']))
       ]
grab_system = ctrl.ControlSystem(rules)
grab_sim = ctrl.ControlSystemSimulation(grab_system)
grab_sim.input['distance'] = 50
grab_sim.input['traffic condition'] = 95
grab_sim.input['demand level'] = 90
grab_sim.input['weather condition'] = 5
grab_sim.input['customer rating'] = 4.8
grab_sim.input['ride punctuality'] = 95
grab_sim.compute()
print(f"Giá cước: {grab_sim.output['price']:.2f}")
print(f"Điểm thưởng: {grab_sim.output['Điểm thưởng']:.2f}")
```

```python
import numpy as np   #2.12
import skfuzzy as fuzz
from skfuzzy import control as ctrl
rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'store rating')
volume = ctrl.Antecedent(np.arange(0, 1000.1, 1), 'sales volume')
profit = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'profit margin')
season = ctrl.Antecedent(np.arange(0, 10.1, 0.1), 'seasonal event')
discount = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'competitor discounts')
chet_khau = ctrl.Consequent(np.arange(0, 70.1, 0.1), 'ti le chiet khau')
rating['low'] = fuzz.trimf(rating.universe, [0, 0, 4])
rating['medium'] = fuzz.trimf(rating.universe, [4, 4.25, 4.5])
rating['high'] = fuzz.trimf(rating.universe, [4.5, 5.0, 5.0])
volume['low'] = fuzz.trimf(volume.universe, [0, 0, 400])
volume['medium'] = fuzz.trimf(volume.universe, [200, 500, 800])
volume['high'] = fuzz.trimf(volume.universe, [600, 1000, 1000])
profit['low'] = fuzz.trimf(profit.universe, [0, 0, 40])
profit['medium'] = fuzz.trimf(profit.universe, [30, 50, 70])
profit['high'] = fuzz.trimf(profit.universe, [60, 100, 100])
season['none'] = fuzz.trimf(season.universe, [0, 0, 3])
season['medium'] = fuzz.trimf(season.universe, [2, 5, 8])
season['high'] = fuzz.trimf(season.universe, [7, 10, 10])
discount['low'] = fuzz.trimf(discount.universe, [0, 0, 40])
discount['medium'] = fuzz.trimf(discount.universe, [30, 50, 70])
discount['high'] = fuzz.trimf(discount.universe, [60, 100, 100])
chet_khau['very low'] = fuzz.trimf(chet_khau.universe, [0, 2.5, 5])
chet_khau['low'] = fuzz.trimf(chet_khau.universe, [5, 7.5, 10])
chet_khau['medium'] = fuzz.trimf(chet_khau.universe, [10, 15, 20])
chet_khau['high'] = fuzz.trimf(chet_khau.universe, [20, 30, 40])
chet_khau['very high'] = fuzz.trimf(chet_khau.universe, [40, 55, 70])
rules = [
    ctrl.Rule(rating['high'] & volume['high'] & profit['high'], chet_khau['very low']),
    ctrl.Rule(rating['low'] & volume['low'] & profit['high'], chet_khau['high']),
    ctrl.Rule(season['high'] & discount['high'], chet_khau['very high']),
    ctrl.Rule(rating['medium'] & volume['medium'] & profit['medium'], chet_khau['medium']),
    ctrl.Rule(discount['low'] & profit['low'] & volume['high'], chet_khau['very low']),
    ctrl.Rule(rating['low'] & season['none'], chet_khau['medium']),
    ctrl.Rule(volume['low'] & profit['low'], chet_khau['very high'])
]
shopee_system = ctrl.ControlSystem(rules)
shopee_sim = ctrl.ControlSystemSimulation(shopee_system)
shopee_sim.input['store rating'] = 4.3
shopee_sim.input['sales volume'] = 350
shopee_sim.input['profit margin'] = 40
shopee_sim.input['seasonal event'] = 10
shopee_sim.input['competitor discounts'] = 70
shopee_sim.compute()
print(f"Tỉ lệ chiết khấu: {shopee_sim.output['ti le chiet khau']:.2f}%")
```

```python
import numpy as np   #2.13
import skfuzzy as fuzz
from skfuzzy import control as ctrl
demand = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'product demand')
pressure = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'CPP')
reputation = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'store reputation')
profit = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'profit margin')
season = ctrl.Antecedent(np.arange(0, 10.1, 0.1), 'seasonal demand')
discount = ctrl.Consequent(np.arange(0, 70.1, 0.1), 'discount percentage')
demand['low'] = fuzz.trimf(demand.universe, [0, 0, 40])
demand['medium'] = fuzz.trimf(demand.universe, [30, 50, 70])
demand['high'] = fuzz.trimf(demand.universe, [60, 100, 100])
pressure['low'] = fuzz.trimf(pressure.universe, [0, 0, 40])
pressure['medium'] = fuzz.trimf(pressure.universe, [30, 50, 70])
pressure['high'] = fuzz.trimf(pressure.universe, [60, 100, 100])
reputation['low'] = fuzz.trimf(reputation.universe, [0, 0, 4.0])
reputation['medium'] = fuzz.trimf(reputation.universe, [3.8, 4.25, 4.6])
reputation['high'] = fuzz.trimf(reputation.universe, [4.4, 5.0, 5.0])
profit['low'] = fuzz.trimf(profit.universe, [0, 0, 30])
profit['medium'] = fuzz.trimf(profit.universe, [20, 45, 70])
profit['high'] = fuzz.trimf(profit.universe, [60, 100, 100])
season['none'] = fuzz.trimf(season.universe, [0, 0, 3])
season['medium'] = fuzz.trimf(season.universe, [2, 5, 8])
season['high'] = fuzz.trimf(season.universe, [7, 10, 10])
discount['very low'] = fuzz.trimf(discount.universe, [0, 2.5, 5])
discount['low'] = fuzz.trimf(discount.universe, [5, 7.5, 10])
discount['medium'] = fuzz.trimf(discount.universe, [10, 20, 30])
discount['high'] = fuzz.trimf(discount.universe, [20, 30, 40])
discount['very high'] = fuzz.trimf(discount.universe, [40, 55, 70])
rules = [
ctrl.Rule(demand['high'] & pressure['low'] & profit['low'], discount['very low']),
ctrl.Rule(demand['low'] & pressure['high'] & profit['high'], discount['high']),
ctrl.Rule(reputation['high'] & profit['medium'] & season['high'], discount['medium']),
ctrl.Rule(pressure['high'] & demand['high'] & profit['high'], discount['very high']),
ctrl.Rule(reputation['low'] & demand['medium'] & profit['low'], discount['medium']),
ctrl.Rule(demand['high'] & season['none'] & pressure['low'], discount['very low']),
ctrl.Rule(profit['high'] & pressure['medium'] & season['medium'], discount['medium']),
ctrl.Rule(profit['high'] & pressure['medium'] & season['high'], discount['medium'])
]
pricing_ctrl = ctrl.ControlSystem(rules)
pricing_sim = ctrl.ControlSystemSimulation(pricing_ctrl)
pricing_sim.input['product demand'] = 70
pricing_sim.input['CPP'] = 50
pricing_sim.input['store reputation'] = 4.2
pricing_sim.input['profit margin'] = 90
pricing_sim.input['seasonal demand'] = 9
pricing_sim.compute()
print(f"Giảm giá: {pricing_sim.output['discount percentage']:.2f}%")
```

```python
import numpy as np   #2.14
import skfuzzy as fuzz
from skfuzzy import control as ctrl
density = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'order density')
urgency = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'delivery urgency')
load = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'driver load')
traffic = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'traffic condition')
profit = ctrl.Antecedent(np.arange(0, 100.1, 0.1), 'profit per delivery')
combine = ctrl.Consequent(np.arange(0, 10.1, 0.1), 'orders to combine')
priority = ctrl.Consequent(np.arange(0, 10.1, 0.1), 'delivery priority')
density['low'] = fuzz.trimf(density.universe, [0, 0, 40])
density['medium'] = fuzz.trimf(density.universe, [30, 50, 70])
density['high'] = fuzz.trimf(density.universe, [60, 100, 100])
urgency['low'] = fuzz.trimf(urgency.universe, [0, 0, 40])
urgency['medium'] = fuzz.trimf(urgency.universe, [30, 50, 70])
urgency['high'] = fuzz.trimf(urgency.universe, [60, 100, 100])
load['low'] = fuzz.trimf(load.universe, [0, 0, 40])
load['medium'] = fuzz.trimf(load.universe, [30, 50, 70])
load['high'] = fuzz.trimf(load.universe, [60, 100, 100])
traffic['low'] = fuzz.trimf(traffic.universe, [0, 0, 40])
traffic['medium'] = fuzz.trimf(traffic.universe, [30, 50, 70])
traffic['high'] = fuzz.trimf(traffic.universe, [60, 100, 100])
profit['low'] = fuzz.trimf(profit.universe, [0, 0, 40])
profit['medium'] = fuzz.trimf(profit.universe, [30, 50, 70])
profit['high'] = fuzz.trimf(profit.universe, [60, 100, 100])
combine['few'] = fuzz.trimf(combine.universe, [0, 2, 4])
combine['some'] = fuzz.trimf(combine.universe, [3, 5, 7])
combine['many'] = fuzz.trimf(combine.universe, [6, 8, 10])
priority['low'] = fuzz.trimf(priority.universe, [0, 2.5, 5])
priority['medium'] = fuzz.trimf(priority.universe, [4, 6, 8])
priority['high'] = fuzz.trimf(priority.universe, [7, 9, 10])
rules = [
ctrl.Rule(density['high'] & load['low'] & traffic['low'], combine['many']),
ctrl.Rule(density['medium'] & traffic['high'] & urgency['medium'], combine['few']),
ctrl.Rule(load['high'] & density['high'] & profit['medium'], combine['some']),
ctrl.Rule(density['low'] & urgency['high'] & traffic['medium'], combine['few']),
ctrl.Rule(profit['high'] & urgency['high'] & traffic['high'], combine['few']),
ctrl.Rule(urgency['high'] & profit['high'], priority['high']),
ctrl.Rule(urgency['medium'] & traffic['medium'], priority['medium']),
ctrl.Rule(urgency['low'] & density['high'] & profit['low'], priority['low']),
ctrl.Rule(density['high'] & load['low'], combine['many']),
]
delivery_ctrl = ctrl.ControlSystem(rules)
delivery_sim = ctrl.ControlSystemSimulation(delivery_ctrl)
delivery_sim.input['order density'] = 85
delivery_sim.input['delivery urgency'] = 50
delivery_sim.input['driver load'] = 15
delivery_sim.input['traffic condition'] = 50
delivery_sim.input['profit per delivery'] = 50
delivery_sim.compute()
print(f"Mức độ kết hợp đơn hàng: {delivery_sim.output['orders to combine']:.2f}")
print(f"Mức độ ưu tiên: {delivery_sim.output['delivery priority']:.2f}")
```
